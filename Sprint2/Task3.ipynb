```
using System;

public class Person
{
    private string name;
    private int age;
    private string gender;
    private string phoneNumber;
    private string email;

    public string Name
    {
        get { return name; }
        set
        {
            if (value == null)
            {
                Console.WriteLine("Имя не может быть null!");
            }
            else
            {
                name = value;
            }
        }
    }

    public int Age
    {
        get { return age; }
        set
        {
            if (value >= 0)
                age = value;
            else
                Console.WriteLine("Возраст не может быть отрицательным!");
        }
    }

    public string Gender
    {
        get { return gender; }
        set
        {
            if (value == null)
            {
                Console.WriteLine("Пол не может быть null!");
            }
            else
            {
                gender = value;
            }
        }
    }

    public string PhoneNumber
    {
        get { return phoneNumber; }
        set
        {
            if (value == null)
            {
                Console.WriteLine("Номер телефона не может быть null!");
            }
            else
            {
                phoneNumber = value;
            }
        }
    }

    public string Email
    {
        get { return email; }
        set
        {
            if (value == null)
            {
                Console.WriteLine("Электронная почта не может быть null!");
            }
            else
            {
                email = value;
            }
        }
    }

    public Person() { }

    public Person(string name, int age, string gender, string phoneNumber, string email)
    {
        Name = name;
        Age = age;
        Gender = gender;
        PhoneNumber = phoneNumber;
        Email = email;
    }

    public virtual string GetInfo()
    {
        return $"Имя: {Name}, Возраст: {Age}, Пол: {Gender}, Телефон: {PhoneNumber}, Электронная почта: {Email}";
    }

    public virtual void SayHello()
    {
        Console.WriteLine($"Приветствую");
    }

    public void IntroduceTo(Person otherPerson)
    {
        Console.WriteLine($"{Name} говорит: Привет, {otherPerson.Name}!");
    }

    public void ChangePassword(string newPassword)
    {
        Console.WriteLine($"Пароль был изменен для {Name}");
    }
}

public class Student : Person
{
    private string university;
    private string course;
    private double gpa;
    private int graduationYear;

    public string University
    {
        get { return university; }
        set
        {
            if (value == null)
            {
                Console.WriteLine("Университет не может быть null!");
            }
            else
            {
                university = value;
            }
        }
    }

    public string Course
    {
        get { return course; }
        set
        {
            if (string.IsNullOrEmpty(value))
                Console.WriteLine("Курс не может быть пустым!");
            else
                course = value;
        }
    }

    public double GPA
    {
        get { return gpa; }
        set
        {
            if (value >= 0 && value <= 4)
                gpa = value;
            else
                Console.WriteLine("Средний балл должен быть между 0 и 4!");
        }
    }

    public int GraduationYear
    {
        get { return graduationYear; }
        set
        {
            if (value >= DateTime.Now.Year)
                graduationYear = value;
            else
                Console.WriteLine("Год выпуска не может быть меньше текущего года!");
        }
    }

    public Student() { }

    public Student(string name, int age, string gender, string university, string course, double gpa, int graduationYear)
        : base(name, age, gender, "", "")
    {
        University = university;
        Course = course;
        GPA = gpa;
        GraduationYear = graduationYear;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}, я учусь на {Course} в {University}, GPA: {GPA}");
    }

    public void Study()
    {
        Console.WriteLine($"{Name} учится на {Course} в {University}. GPA: {GPA}, год выпуска: {GraduationYear}");
    }
}

public class Employee : Person
{
    private string company;
    private double salary;
    private string position;
    private DateTime hireDate;

    public string Company
    {
        get { return company; }
        set
        {
            if (value == null)
            {
                Console.WriteLine("Компания не может быть null!");
            }
            else
            {
                company = value;
            }
        }
    }

    public double Salary
    {
        get { return salary; }
        set
        {
            if (value >= 0)
                salary = value;
            else
                Console.WriteLine("Зарплата не может быть отрицательной!");
        }
    }

    public string Position
    {
        get { return position; }
        set
        {
            if (string.IsNullOrEmpty(value))
                Console.WriteLine("Должность не может быть пустой!");
            else
                position = value;
        }
    }

    public DateTime HireDate
    {
        get { return hireDate; }
        set
        {
            if (value <= DateTime.Now)
                hireDate = value;
            else
                Console.WriteLine("Дата приема на работу не может быть в будущем!");
        }
    }

    public Employee() { }

    public Employee(string name, int age, string gender, string company, double salary, string position, DateTime hireDate)
        : base(name, age, gender, "", "")
    {
        Company = company;
        Salary = salary;
        Position = position;
        HireDate = hireDate;
    }

    public override string GetInfo()
    {
        return $"{base.GetInfo()}, Компания: {Company}, Должность: {Position}, Зарплата: {Salary:C}, Дата приема на работу: {HireDate.ToShortDateString()}";
    }

    public void Work()
    {
        Console.WriteLine($"{Name} работает в {Company} на должности {Position}. Зарплата: {Salary:C}, Дата приема на работу: {HireDate.ToShortDateString()}");
    }
}

public class Teacher : Person
{
    private string subject;
    private int experience;
    private int teachingHours;
    private string researchArea;

    public string Subject
    {
        get { return subject; }
        set
        {
            if (value == null)
            {
                Console.WriteLine("Предмет не может быть null!");
            }
            else
            {
                subject = value;
            }
        }
    }

    public int Experience
    {
        get { return experience; }
        set
        {
            if (value >= 0)
                experience = value;
            else
                Console.WriteLine("Опыт не может быть отрицательным!");
        }
    }

    public int TeachingHours
    {
        get { return teachingHours; }
        set
        {
            if (value >= 0)
                teachingHours = value;
            else
                Console.WriteLine("Часы преподавания не могут быть отрицательными!");
        }
    }

    public string ResearchArea
    {
        get { return researchArea; }
        set
        {
            if (string.IsNullOrEmpty(value))
                Console.WriteLine("Область исследования не может быть пустой!");
            else
                researchArea = value;
        }
    }

    public Teacher() { }

    public Teacher(string name, int age, string gender, string subject, int experience, int teachingHours, string researchArea)
        : base(name, age, gender, "", "")
    {
        Subject = subject;
        Experience = experience;
        TeachingHours = teachingHours;
        ResearchArea = researchArea;
    }

    public override void SayHello()
    {
        Console.WriteLine($"Привет, меня зовут {Name}, я преподаю дисциплину {Subject} уже {Experience} лет. Исследую {ResearchArea}");
    }

    public void PublishPaper()
    {
        Console.WriteLine($"{Name} опубликовал научную статью по теме: {ResearchArea}. Преподает {Subject}, часы преподавания: {TeachingHours}.");
    }
}

```

```

Person person = new Person("Игорь Иванов", 35, "Мужчина", "123-456-7890", "igor@example.com");
Student student = new Student("Алиса Жирикова", 22, "Женщина", "Тюменский индустриальный университет", "Программирование", 3.8, 2025);
Employee employee = new Employee("Арсений Васенин", 25, "Мужчина", "ООО \"СуперПупер\"", 50000, "Программист", new DateTime(2022, 6, 1));
Teacher teacher = new Teacher("Ольга Рукова", 40, "Женщина", "Математика", 20, 30, "Математическое моделирование");

person.IntroduceTo(student);
student.SayHello();
employee.Work();
teacher.PublishPaper();
```
